### Task 3 - Create the fact_transactions with incremental update
- create table
- Load initial transactions
- Perform incremental update with data from parqueut

```python
%sql
    
    
-- create table
CREATE OR REPLACE TABLE IDENTIFIER(:catalog_name || '.2_silver.fact_transactions') (
  transaction_id STRING NOT NULL,
  account_id STRING,
  customer_id STRING,
  transaction_type STRING,
  amount DECIMAL(15,2),
  currency STRING,
  transaction_date DATE,
  description STRING
)
USING DELTA
CLUSTER BY (customer_id, transaction_date)
TBLPROPERTIES ('delta.enableChangeDataFeed' = 'true');

-- Enable column defaults feature before setting default value
ALTER TABLE IDENTIFIER(:catalog_name || '.2_silver.fact_transactions')
  SET TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE IDENTIFIER(:catalog_name || '.2_silver.fact_transactions')
  ALTER COLUMN currency SET DEFAULT 'CAD';
```
