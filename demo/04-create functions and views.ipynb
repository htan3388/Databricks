### Task 4: create function and view
- Function: kpi
- View: vw_transactions

```python
%sql
    
-- # create a function
CREATE OR REPLACE FUNCTION IDENTIFIER(:catalog_name || '.2_silver.kpi') (amt DOUBLE)
RETURNS STRING
LANGUAGE SQL
RETURN
  CASE
    WHEN amt < 1000 THEN 'Low'    
    ELSE 'High'
  END;

 
-- # create a standard view
CREATE OR REPLACE VIEW IDENTIFIER(:catalog_name || '.2_silver.vw_transactions')
AS
SELECT * FROM IDENTIFIER(:catalog_name || '.2_silver.fact_transactions');
-- Note: function calls in expressions (dev.2_silver.kpi) cannot be parameterized with IDENTIFIER()

/*
-- # Create a materialized view for customer statistics
CREATE  MATERIALIZED VIEW IDENTIFIER(:catalog_name || '.3_gold.vw_customer_stats')
AS
SELECT c.customer_id, c.full_name, COUNT(*) AS total_transactions, SUM(t.amount) AS total_amount
FROM IDENTIFIER(:catalog_name || '.2_silver.fact_transactions') t
JOIN IDENTIFIER(:catalog_name || '.2_silver.dim_customer') c
  ON t.customer_id = c.customer_id
GROUP BY c.customer_id, c.full_name;

SELECT * FROM IDENTIFIER(:catalog_name || '.3_gold.vw_customer_stats')
ORDER BY customer_id;
*/
```
